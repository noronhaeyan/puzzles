**Suggested workflow**

1. Start with Part 1 and reason about the boundary cases.

3. Reuse ideas only when the rules genuinely match; Part 3 changes the state transition, and Part 4 changes the infection condition.
4. Use the final test cell to run the complete score.

```python
# Shared interview harness. Run this cell once before running any test cell.
import copy
import time

TIME_LIMIT = 10.0


def score_cases(function, cases):
    passed = 0
    print(f"Testing {function.__name__}")
    for name, args, expected in cases:
        started = time.perf_counter()
        try:
            actual = function(*copy.deepcopy(args))
            elapsed = time.perf_counter() - started
            if actual == expected and elapsed <= TIME_LIMIT:
                passed += 1
                print(f"  PASS  {name}")
            elif actual != expected:
                print(f"  FAIL  {name}: expected {expected!r}, got {actual!r}")
            else:
                print(f"  FAIL  {name}: exceeded {TIME_LIMIT:g}s")
        except Exception as error:
            print(f"  ERROR {name}: {type(error).__name__}: {error}")
    print(f"Score: {passed}/{len(cases)}")
    return passed, len(cases)
```
